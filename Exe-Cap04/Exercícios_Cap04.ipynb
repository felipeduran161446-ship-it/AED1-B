Exercício 13 — Comparando sua solução com a IA

Objetivo: Comparar uma solução própria, escrita antes de
consultar a IA, com a solução gerada por uma IA.

Enunciado: Antes de usar qualquer IA, escreva o seu próprio algoritmo para "calcular o valor de uma
compra com desconto". Depois, peça a uma IA que resolva o mesmo problema e compare
as duas soluções usando os critérios abaixo

### Minha Solução (Como Usuário):

**Nome do Algoritmo:** Cálculo de Preço com Desconto

**Entradas:**
1.  `preco_original`: O valor inicial do produto ou da compra.
2.  `percentual_desconto`: A porcentagem de desconto a ser aplicada (ex: 10 para 10%).

**Passos:**
1.  **Converter Porcentagem:** Dividir o `percentual_desconto` por 100 para obter o valor decimal do desconto.
    *   `desconto_decimal = percentual_desconto / 100`
2.  **Calcular Valor do Desconto:** Multiplicar o `preco_original` pelo `desconto_decimal`.
    *   `valor_do_desconto = preco_original * desconto_decimal`
3.  **Calcular Preço Final:** Subtrair o `valor_do_desconto` do `preco_original`.
    *   `preco_final = preco_original - valor_do_desconto`

**Saída:**
-   `preco_final`: O valor da compra após a aplicação do desconto.

### Solução Gerada pela IA:

**Nome do Algoritmo:** Algoritmo Otimizado de Desconto para Compras

**Entradas:**
1.  `valor_base_compra`: O valor total da compra antes de qualquer desconto. (Deve ser um número não negativo).
2.  `taxa_desconto_percentual`: A taxa de desconto a ser aplicada, expressa como uma porcentagem (ex: 15 para 15%). (Deve ser um número entre 0 e 100).

**Passos:**
1.  **Validação de Entradas:**
    *   Verificar se `valor_base_compra` é menor que 0. Se sim, retornar erro ("Valor base inválido").
    *   Verificar se `taxa_desconto_percentual` é menor que 0 ou maior que 100. Se sim, retornar erro ("Taxa de desconto inválida").
2.  **Cálculo do Fator Multiplicador:**
    *   `fator_desconto = (100 - taxa_desconto_percentual) / 100`
3.  **Cálculo do Valor Final:**
    *   `valor_final_com_desconto = valor_base_compra * fator_desconto`

**Saída:**
-   `valor_final_com_desconto`: O preço da compra após a aplicação do desconto. Em caso de entradas inválidas, retorna a mensagem de erro apropriada.

### Critérios de Comparação para as Soluções:

Ao comparar os dois algoritmos, você pode considerar os seguintes critérios:

1.  **Clareza e Compreensibilidade:** Qual algoritmo é mais fácil de entender rapidamente?
2.  **Precisão:** Ambos chegam ao resultado correto?
3.  **Robustez:** Qual algoritmo lida melhor com entradas inesperadas ou inválidas (por exemplo, um desconto negativo ou um valor de compra irreal)?
4.  **Eficiência:** Embora para este problema simples não seja um fator crítico, qual algoritmo parece mais direto ou tem menos passos de cálculo redundantes?
5.  **Generalização:** Qual deles poderia ser mais facilmente adaptado ou estendido para cenários mais complexos (como múltiplos descontos, impostos, etc.)?
6.  **Formato:** Como a estrutura e a apresentação de cada algoritmo contribuem para sua utilidade?

```python
### Implementação da Minha Solução (Como Usuário)

def calcular_desconto_usuario(preco_original, percentual_desconto):
    """
    Calcula o valor final de uma compra com desconto usando a lógica do usuário.

    Args:
        preco_original (float): O valor inicial do produto ou da compra.
        percentual_desconto (float): A porcentagem de desconto a ser aplicada (ex: 10 para 10%).

    Returns:
        float: O valor da compra após a aplicação do desconto.
    """
    # 1. Converter Porcentagem
    desconto_decimal = percentual_desconto / 100

    # 2. Calcular Valor do Desconto
    valor_do_desconto = preco_original * desconto_decimal

    # 3. Calcular Preço Final
    preco_final = preco_original - valor_do_desconto

    return preco_final

# Exemplo de uso:
preco = 150.0
desconto = 20
preco_com_desconto_usuario = calcular_desconto_usuario(preco, desconto)
print(f"Preço original: R${preco:.2f}")
print(f"Desconto: {desconto}%")
print(f"Preço final (Minha Solução): R${preco_com_desconto_usuario:.2f}")
```

```python
### Implementação da Solução Gerada pela IA

def calcular_desconto_ia(valor_base_compra, taxa_desconto_percentual):
    """
    Calcula o valor final de uma compra com desconto usando a lógica otimizada da IA.

    Args:
        valor_base_compra (float): O valor total da compra antes de qualquer desconto.
        taxa_desconto_percentual (float): A taxa de desconto a ser aplicada (ex: 15 para 15%).

    Returns:
        float or str: O preço da compra após o desconto, ou uma mensagem de erro.
    """
    # 1. Validação de Entradas
    if valor_base_compra < 0:
        return "Erro: Valor base de compra inválido (não pode ser negativo)."
    if not (0 <= taxa_desconto_percentual <= 100):
        return "Erro: Taxa de desconto inválida (deve estar entre 0 e 100%)."

    # 2. Cálculo do Fator Multiplicador
    fator_desconto = (100 - taxa_desconto_percentual) / 100

    # 3. Cálculo do Valor Final
    valor_final_com_desconto = valor_base_compra * fator_desconto

    return valor_final_com_desconto

# Exemplo de uso:
preco = 200.0
desconto = 15
preco_com_desconto_ia = calcular_desconto_ia(preco, desconto)
print(f"\nPreço original: R${preco:.2f}")
print(f"Desconto: {desconto}%")
print(f"Preço final (Solução IA): R${preco_com_desconto_ia:.2f}")

# Exemplo com entrada inválida:
preco_invalido = -50.0
desconto_invalido = 110
print(f"\nTeste com preço inválido: {calcular_desconto_ia(preco_invalido, 10)}")
print(f"Teste com desconto inválido: {calcular_desconto_ia(100, desconto_invalido)}")
```
