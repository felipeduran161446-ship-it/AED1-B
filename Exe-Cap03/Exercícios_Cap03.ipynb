Exercício 1 — Bits e bytes
Objetivo: Diferenciar bit e byte e converter uma quantidade de bytes
em bits.

Enunciado: Explique, com suas próprias palavras, a diferença entre bit e byte. Depois, responda:
quantos bits existem em 5 bytes?

Um **bit** (binary digit) é a menor unidade de informação em computação, representando um valor binário de 0 ou 1. É como um interruptor que pode estar ligado ou desligado.

Um **byte** é uma unidade de informação digital que consiste em 8 bits. É a unidade básica de armazenamento e processamento na maioria dos computadores, capaz de representar um caractere (como uma letra, número ou símbolo). Pense em um byte como um grupo de 8 interruptores de luz.

Para responder à pergunta:

Sabendo que 1 byte = 8 bits, para encontrar quantos bits existem em 5 bytes, multiplicamos 5 por 8.

```python
bits_per_byte = 8
num_bytes = 5

total_bits = num_bytes * bits_per_byte

print(f"Em {num_bytes} bytes, existem {total_bits} bits.")
```

Exercício 4 — Decimal para binário no papel

Objetivo: Converter números decimais para binário pelo método das
divisões sucessivas por 2.

Enunciado: Converta os números 13, 42, 7, 100 e 255 para binário usando divisões sucessivas por 2.
Depois, verifique no Colab com bin().

Para converter um número decimal para binário usando divisões sucessivas por 2, siga estes passos:
1. Divida o número decimal por 2.
2. Anote o resto (que será 0 ou 1).
3. Use o quociente como o novo número a ser dividido e repita os passos 1 e 2.
4. Continue até que o quociente seja 0.
5. O número binário é formado pelos restos lidos de baixo para cima (do último resto ao primeiro).

Vamos aplicar este método para cada número e depois verificar com a função `bin()` no Colab.

```python
numbers_to_convert = [13, 42, 7, 100, 255]

print("Verificação de Decimal para Binário:")
for num in numbers_to_convert:
    binary_representation = bin(num)
    print(f"O número decimal {num} em binário é: {binary_representation} (ignorar o prefixo '0b')")
```

Exercício 8 — Encontre o erro de unidade

Objetivo: Identificar um fator que faltou em um cálculo de
tamanho de imagem.

Enunciado: Um estudante calculou o tamanho de uma imagem RGB assim:

largura = 1920

altura = 1080

tamanho = largura * altura

print(tamanho)

Explique o que está faltando no cálculo.

O cálculo do estudante `largura * altura` nos dá apenas o **número total de pixels** na imagem.

Para calcular o **tamanho real da imagem em bytes**, é preciso considerar mais alguns fatores:

1.  **Número de Canais de Cor:** Uma imagem RGB (Red, Green, Blue) possui 3 canais de cor. Cada pixel tem informações para Vermelho, Verde e Azul. Portanto, o número de pixels deve ser multiplicado por 3 (canais).
2.  **Profundidade de Cor (Bit Depth):** Cada canal de cor geralmente armazena sua informação usando um certo número de bits. O mais comum é 8 bits por canal (o que é igual a 1 byte por canal). Portanto, o resultado deve ser multiplicado pelo número de bytes por canal (geralmente 1).

**Cálculo correto:**

Tamanho da imagem = `largura * altura * número de canais * bytes por canal`

Por exemplo, para uma imagem RGB de 8 bits por canal:
Tamanho = `1920 * 1080 * 3 canais * 1 byte/canal`

```python
largura = 1920
altura = 1080
canais_de_cor = 3 # Para imagens RGB (Vermelho, Verde, Azul)
bits_por_canal = 8 # Profundidade de cor comum, 8 bits por canal

# Convertendo bits por canal para bytes por canal
bytes_por_canal = bits_por_canal / 8

# Cálculo correto do tamanho da imagem em bytes
tamanho_total_bytes = largura * altura * canais_de_cor * bytes_por_canal

print(f"O número de pixels é: {largura * altura}")
print(f"O tamanho CORRETO da imagem é: {tamanho_total_bytes} bytes")
```

Exercício 11 — Calculadora de imagem no Colab

Objetivo: Construir um notebook que calcule o tamanho bruto de uma
imagem RGB em bytes, KB e MB.

Enunciado: Crie um notebook no Colab que calcule o tamanho bruto de uma imagem RGB. O notebook
deve permitir alterar largura e altura manualmente no código, e mostrar o resultado em
bytes, KB e MB.

### Calculadora de Tamanho de Imagem RGB

Nesta seção, você pode ajustar a largura e altura da imagem para calcular seu tamanho bruto em Bytes, Kilobytes (KB) e Megabytes (MB).

```python
# --- Parâmetros da Imagem (ajuste aqui) ---
largura = 1920  # Exemplo: Largura em pixels
altura = 1080   # Exemplo: Altura em pixels

# --- Características da Imagem RGB (geralmente fixas) ---
canais_de_cor = 3  # Para RGB: Vermelho, Verde, Azul
bits_por_canal = 8 # Profundidade de cor comum: 8 bits por canal

# --- Cálculos ---

# Converter bits por canal para bytes por canal
bytes_por_canal = bits_por_canal / 8

# Tamanho total em Bytes
tamanho_bytes = largura * altura * canais_de_cor * bytes_por_canal

# Tamanho em Kilobytes (KB)
tamanho_kb = tamanho_bytes / 1024

# Tamanho em Megabytes (MB)
tamanho_mb = tamanho_kb / 1024

# --- Exibição dos Resultados ---
print(f"Dimensões da imagem: {largura}x{altura} pixels")
print(f"Canais de cor (RGB): {canais_de_cor}")
print(f"Profundidade de cor por canal: {bits_por_canal} bits")
print("\n---")
print(f"Tamanho bruto da imagem:")
print(f"  Bytes: {tamanho_bytes:,.2f} Bytes")
print(f"  Kilobytes (KB): {tamanho_kb:,.2f} KB")
print(f"  Megabytes (MB): {tamanho_mb:,.2f} MB")
```

Exercício 14 — Peça explicação à IA e audite
Objetivo: Auditar uma explicação de IA sobre a conversão de 42

Enunciado: Use o prompt abaixo para pedir a uma IA que explique a conversão de 42 para binário.
Depois, responda às quatro perguntas de auditoria.

Explique para um estudante iniciante como converter o número
decimal 42 para binário usando divisões sucessivas por 2.
Mostre cada etapa, explique por que os restos são lidos de
baixo para cima e inclua uma verificação somando potências
de 2.


Olá, estudante! Vamos aprender a converter o número decimal **42** para binário, que é a linguagem que os computadores entendem, usando um método chamado **divisões sucessivas por 2**.

### Passo a Passo: Conversão de 42 para Binário

1.  **Começamos com o número 42 e dividimos por 2:**
    *   `42 ÷ 2 = 21` com **resto 0**

2.  **Pegamos o quociente (21) e dividimos novamente por 2:**
    *   `21 ÷ 2 = 10` com **resto 1**

3.  **Continuamos com o quociente (10) e dividimos por 2:**
    *   `10 ÷ 2 = 5` com **resto 0**

4.  **Novamente, pegamos o quociente (5) e dividimos por 2:**
    *   `5 ÷ 2 = 2` com **resto 1**

5.  **Mais uma vez, pegamos o quociente (2) e dividimos por 2:**
    *   `2 ÷ 2 = 1` com **resto 0**

6.  **E a última divisão, pegamos o quociente (1) e dividimos por 2:**
    *   `1 ÷ 2 = 0` com **resto 1**

Chegamos ao quociente 0, então paramos as divisões. Agora, temos uma lista de restos:
**0, 1, 0, 1, 0, 1** (na ordem em que foram gerados).

### Por que lemos os restos de baixo para cima?

Imagine que cada divisão por 2 nos diz se o número original tinha um '1' ou um '0' em uma determinada posição binária. Quando você divide um número por 2, o resto (0 ou 1) é o dígito binário menos significativo para aquele *estágio* da divisão. O último quociente (o 1 final no nosso exemplo) é o dígito binário mais significativo do número completo.

Pense assim: A primeira divisão nos dá o 'valor de unidade' (o bit mais à direita) em binário. A próxima divisão nos dá o 'valor de dois' (o segundo bit da direita para a esquerda), e assim por diante. Se lermos de cima para baixo, estaríamos começando com o bit de menor valor e terminando com o de maior valor, o que resultaria no número binário invertido. Para obter a representação correta (do bit mais significativo para o menos significativo), precisamos ler os restos na ordem inversa da qual foram gerados – ou seja, **de baixo para cima**.

No nosso exemplo, lendo os restos de baixo para cima (do último resto para o primeiro), obtemos:

**101010**

Então, o número decimal 42 é **101010** em binário.

### Verificação: Somando potências de 2

Para verificar se nossa conversão está correta, podemos somar as potências de 2 onde o bit binário correspondente é 1. Lembre-se que em binário, cada posição representa uma potência de 2, começando da direita para a esquerda com 2^0, 2^1, 2^2, e assim por diante.

Nosso número binário é **101010**:

*   `1` x 2^5 = 1 x 32 = 32
*   `0` x 2^4 = 0 x 16 = 0
*   `1` x 2^3 = 1 x 8 = 8
*   `0` x 2^2 = 0 x 4 = 0
*   `1` x 2^1 = 1 x 2 = 2
*   `0` x 2^0 = 0 x 1 = 0

Agora, somamos esses valores:
`32 + 0 + 8 + 0 + 2 + 0 = 42`

Como a soma é 42, nossa conversão está **correta**!
