**Controle de Entrada de Cinema**

```python
idade = int(input("Qual a sua idade"))
carteira = float(input("Quanto você tem na sua carteira"))

if idade >= 16:
         print("Entrada permitida")

else:
         print("Entrada proibida")

if carteira >= 30.00:
         print("Ingresso comprado! Bom filme")

else:
         print("Saldo insuficinete para o ingresso")

print("Foi bom te atender!!")
```
