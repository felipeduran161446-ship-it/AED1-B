```python
palavra = 'FANTASMA'
print('Letra | Decimal | Binário')
print('-' * 30)
for letra in palavra:
 codigo = ord(letra)
 binario = format(codigo, '08b')
 print(f'{letra:5} | {codigo:7} | {binario}')
```

```python
nome = 'Felipe Duran'
print('Caractere | Decimal | Binário')
print('-' * 35)
for c in nome:
 codigo = ord(c)
 binario = format(codigo, '08b')
 print(f'{repr(c):9} | {codigo:7} | {binario}')
```
