Exercício 1 — Identificando símbolos

Objetivo: Associar cada ação ao símbolo correto de fluxograma.

Enunciado:Associe cada ação abaixo ao símbolo correto de fluxograma: a) ler a idade do usuário; b)
calcular media = (n1 + n2) / 2; c) verificar se idade >= 18; d) marcar o início do
algoritmo; e) exibir "Cadastro aprovado".


## Solução — Exercício 1

Os símbolos de fluxograma mais comuns para estas ações são:

*   **Terminal (Cápsula/Retângulo Arredondado):** Indica o início ou fim de um processo.
    *   `Símbolo:` <img src="https://upload.wikimedia.org/wikipedia/commons/thumb/e/e0/Flowchart_terminal.svg/100px-Flowchart_terminal.svg.png" alt="Símbolo Terminal" width="50" height="30">
*   **Processo (Retângulo):** Representa uma etapa de processamento ou uma ação.
    *   `Símbolo:` <img src="https://upload.wikimedia.org/wikipedia/commons/thumb/d/d4/Flowchart_process.svg/100px-Flowchart_process.svg.png" alt="Símbolo Processo" width="50" height="30">
*   **Entrada/Saída (Paralelogramo):** Indica operações de entrada (ler dados) ou saída (exibir informações).
    *   `Símbolo:` <img src="https://upload.wikimedia.org/wikipedia/commons/thumb/0/05/Flowchart_input-output.svg/100px-Flowchart_input-output.svg.png" alt="Símbolo Entrada/Saída" width="50" height="30">
*   **Decisão (Losango):** Representa um ponto onde é feita uma pergunta, e a resposta determina o caminho a seguir.
    *   `Símbolo:` <img src="https://upload.wikimedia.org/wikipedia/commons/thumb/2/23/Flowchart_decision.svg/100px-Flowchart_decision.svg.png" alt="Símbolo Decisão" width="50" height="30">

Aqui estão as associações:

a) **Ler a idade do usuário:** `Entrada/Saída` (Paralelogramo)

b) **Calcular media = (n1 + n2) / 2:** `Processo` (Retângulo)

c) **Verificar se idade >= 18:** `Decisão` (Losango)

d) **Marcar o início do algoritmo:** `Terminal` (Cápsula/Retângulo Arredondado)

e) **Exibir "Cadastro aprovado":** `Entrada/Saída` (Paralelogramo)

Exercício 10 — Classificação de IMC

Objetivo: Encadear três losangos para classificar o IMC em quatro
categorias.

Enunciado:Desenhe um fluxograma que calcule o IMC (peso / altura²) e o classifique em: abaixo
do peso (< 18,5), peso normal (18,5 a 24,9), sobrepeso (25 a 29,9) ou obesidade (>= 30).


```mermaid
flowchart TD
    start[Início] --> input_peso[Ler Peso (kg)]
    input_peso --> input_altura[Ler Altura (m)]
    input_altura --> calc_imc[IMC = Peso / (Altura * Altura)]

    calc_imc --> decision1{IMC < 18.5?}
    decision1 -- Sim --> result_abaixo[Exibir "Abaixo do peso"]
    decision1 -- Não --> decision2{IMC < 25?}

    decision2 -- Sim --> result_normal[Exibir "Peso normal"]
    decision2 -- Não --> decision3{IMC < 30?}

    decision3 -- Sim --> result_sobrepeso[Exibir "Sobrepeso"]
    decision3 -- Não --> result_obesidade[Exibir "Obesidade"]

    result_abaixo --> end[Fim]
    result_normal --> end
    result_sobrepeso --> end
    result_obesidade --> end
```
