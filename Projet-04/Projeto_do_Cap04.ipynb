**Projeto do capítulo: Manual de instruções para um robô sem bom senso**

Este projeto é aberto: você escolhe uma tarefa simples do cotidiano e escreve um manual
de instruções tão detalhado que até um robô sem nenhum bom senso conseguiria executálo.


TAREFA ESCOLHIDA: ESCOVAR OS DENTES
PRIMEIRA VERSÃO (ANTES DE IDENTIFICAR PROBLEMAS)
ALGORITMO escovar_os_dentes — versão inicial

Pegue a escova de dentes.

Abra a pasta de dente.

Coloque pasta na escova.

Molhe a escova.

Escove os dentes.

Enxágue a boca.

Guarde a escova.

Essa versão parece óbvia para um humano, mas para um robô sem senso comum é um desastre total: qual escova? Onde está? O que é "pasta"? Quanto colocar? O que são "dentes"? Como escovar? Por quanto tempo? O que é "enxaguar"? Com o quê? Guardar onde?

AMBIGUIDADES IDENTIFICADAS (ANÁLISE PRÉVIA)
Passo	Problema para um robô literal
"Pegue a escova"	Qual escova? (há várias? É nova? Usada?). Onde ela está? (gaveta, copo, chão?). Está inteira? Tem cerdas?
"Abra a pasta de dente"	Onde está a pasta? Como abrir? (tampa rosqueável, flip-top, lacre?). Está vazia?
"Coloque pasta na escova"	Quanto? (1mm, 5cm, tubo inteiro?). Onde exatamente nas cerdas? Em cima ou entre elas?
"Molhe a escova"	Com o quê? (água da torneira, água mineral, suco?). Quanto tempo? Até que ponto?
"Escove os dentes"	Quais dentes? (todos? só os da frente?). Movimento? (circular, vertical, horizontal?). Pressão? (leve, forte?). Por quanto tempo? (2 segundos ou 2 minutos?). E a língua?
"Enxágue a boca"	Com o quê? (água? enxaguante?). Quanto? (um gole, um litro?). Cuspir onde? (pia, chão, copo?).
"Guarde a escova"	Onde? (no copo, na gaveta, de cabeça para cima ou para baixo?). A pasta deve ser fechada antes?
VERSÃO FINAL (MANUAL PARA ROBÔ LITERAL)
MANUAL DE INSTRUÇÕES PARA ESCOVAÇÃO DE DENTES
Versão 1.0 - Robô sem bom senso

PRÉ-CONDIÇÕES OBRIGATÓRIAS (VERIFICAR ANTES DE COMEÇAR):

O robô deve estar em posição vertical (em pé) diante de uma pia com torneira funcionante.

Deve haver um espelho à frente (opcional, mas recomendado para verificação visual).

O robô deve possuir um sistema de visão computacional e sensores de pressão/força nos atuadores.

PASSO 1: LOCALIZAÇÃO E SELEÇÃO DA ESCOVA DE DENTES

1.1. Realizar varredura visual de 360° no raio de 2 metros.
1.2. Identificar todos os objetos cilíndricos alongados com cerdas em uma extremidade (cor predominante: branca, azul ou verde; comprimento entre 15 e 20 cm).
1.3. Se houver múltiplas escovas, selecionar a que estiver mais próxima do robô.
1.4. Se não houver escova visível, emitir alerta sonoro "ESCOVA NÃO ENCONTRADA" e abortar missão.
1.5. Estender o braço robótico em direção à escova.
1.6. Agarrar a escova pelo cabo (extremidade oposta às cerdas) com pressão de 2,5 N (newtons) — suficiente para segurar sem esmagar.
1.7. Levantar a escova a 30 cm da superfície.

PASSO 2: LOCALIZAÇÃO E ABERTURA DA PASTA DE DENTES

2.1. Realizar varredura visual para localizar um tubo deformável (formato retangular com bordas arredondadas) com rótulo contendo a palavra "DENTE" ou "FLÚOR" (usar OCR).
2.2. Se não encontrada em 10 segundos, verificar gavetas superiores (abrir com força de 5 N).
2.3. Pegar o tubo com a outra mão (ou com o mesmo braço, se houver pinça rotatória).
2.4. Posicionar o tubo na vertical, com a tampa para cima.
2.5. Identificar o tipo de tampa:

Se for rosqueável: girar no sentido anti-horário com torque de 0,8 N.m por 3 voltas completas.

Se for flip-top (tampa de abrir): aplicar pressão de 3 N na parte inferior da tampa para destravá-la e abrir 180°.

Se for lacre metálico (primeiro uso): perfurar com a ponta da tampa (aplicar 10 N de pressão) e remover o lacre com pinça.
2.6. Se a pasta estiver vazia (peso < 10g), emitir alerta "PASTA VAZIA" e abortar.

PASSO 3: APLICAÇÃO DA PASTA NA ESCOVA (QUANTIDADE PRECISA)

3.1. Posicionar a escova com as cerdas voltadas para cima, inclinação de 15° em relação à horizontal.
3.2. Posicionar o bocal do tubo de pasta a 2 mm acima do centro das cerdas.
3.3. Aplicar pressão constante de 2 N no corpo do tubo enquanto desliza o bocal em linha reta da extremidade proximal (próxima ao cabo) até a extremidade distal (ponta) das cerdas, cobrindo exatamente 80% do comprimento das cerdas.
3.4. A quantidade extraída deve ser equivalente a um cilindro de pasta com diâmetro de 5 mm e comprimento de 15 mm (aproximadamente 1,2 cm³ — tamanho de uma ervilha).
3.5. Caso a pasta escorra para o cabo, limpar com papel absorvente (força de 1 N, movimento único).

PASSO 4: UMIDIFICAÇÃO DA ESCOVA (MOLHAR)

4.1. Localizar a torneira da pia (objeto metálico com alavanca ou duas volutas).
4.2. Verificar se há um copo ou recipiente embaixo da torneira (remover se houver).
4.3. Abrir a torneira:

Se for alavanca única: mover para a posição central (água morna) e levantar 45°.

Se for duas volutas: girar a voluta da água fria no sentido anti-horário por 2 voltas completas.
4.4. Aguardar a água fluir por 2 segundos (para sair a água estagnada).
4.5. Posicionar as cerdas da escova sob o fluxo de água, a 3 cm do bocal da torneira.
4.6. Manter sob a água por exatamente 1,5 segundos (tempo suficiente para umedecer sem encharcar).
4.7. Fechar a torneira (movimento inverso ao de abertura).
4.8. Sacudir a escova uma única vez com amplitude de 10 cm para remover excesso de água (cuidado para não projetar pasta para fora).

PASSO 5: ESCOVAÇÃO DOS DENTES (PROCEDIMENTO DETALHADO)

5.1. Abrir a boca em um ângulo de 40° (medido entre o maxilar superior e inferior).
5.2. Inserir a escova na cavidade oral com as cerdas voltadas para os dentes superiores.

5.3. DIVIDIR A BOCA EM 4 QUADRANTES (tempo total: 120 segundos):

Quadrante 1: Dentes superiores direito (30 segundos)

Quadrante 2: Dentes superiores esquerdo (30 segundos)

Quadrante 3: Dentes inferiores direito (30 segundos)

Quadrante 4: Dentes inferiores esquerdo (30 segundos)

5.4. PARA CADA QUADRANTE, EXECUTAR ESTA SEQUÊNCIA:

Posicionar as cerdas em um ângulo de 45° em relação à gengiva (linha entre dente e gengiva).

Aplicar pressão de 1,5 N (força leve — equivalente a segurar um ovo sem quebrar).

Realizar movimentos circulares no sentido horário com raio de 3 mm, frequência de 2 Hz (2 ciclos por segundo).

Após 10 segundos de movimentos circulares, escovar a superfície mastigatória (parte de cima do dente) com movimentos de vaivém (frente-trás) por 10 segundos, curso de 5 mm.

Por fim, escovar a superfície interna (parte de trás dos dentes) com movimentos verticais (de cima para baixo) por 10 segundos.

NÃO ESCOVAR A GENGIVA — manter as cerdas sempre sobre o esmalte do dente (usar sensor de distância < 1 mm da gengiva para parar).

5.5. APÓS OS 4 QUADRANTES (120 segundos):

Escovar a língua (opcional, mas recomendado): posicionar as cerdas sobre a superfície da língua e fazer 5 movimentos suaves (pressão de 0,5 N) da parte de trás para a frente.

5.6. Retirar a escova da boca lentamente (para não bater nos dentes frontais).

PASSO 6: ENXÁGUE DA BOCA (COM ÁGUA)

6.1. Abrir a torneira novamente (mesmo procedimento do Passo 4.3).
6.2. Encher um copo (previamente posicionado ao lado) com 150 ml de água (medido por sensor de volume ou por tempo: 5 segundos com fluxo médio de 30 ml/s).
6.3. Fechar a torneira.
6.4. Levar o copo à boca com o braço robótico.
6.5. Inclinar a cabeça para trás em 30° (para não derramar).
6.6. Despejar 50 ml de água na cavidade oral (3 goles pequenos).
6.7. Fechar a boca e manter os lábios selados.
6.8. Fazer movimentos de "bochecho": inflar as bochechas alternadamente (ciclo de 2 segundos) por 10 segundos.
6.9. Inclinar a cabeça para frente e para baixo (45°).
6.10. Abrir a boca e deixar a água escorrer para a pia (não cuspir com força — só deixar escorrer para evitar respingos).
6.11. Repetir os passos 6.6 a 6.10 mais duas vezes (total de 3 enxágues com 50 ml cada).
6.12. Descartar o copo (ou colocá-lo de volta no lugar com abertura para cima).

PASSO 7: LIMPEZA E ARMAZENAMENTO DA ESCOVA

7.1. Lavar a escova: colocar as cerdas sob a torneira aberta por 3 segundos (água fria) para remover resíduos de pasta.
7.2. Sacudir a escova com 3 movimentos bruscos (amplitude de 15 cm) sobre a pia.
7.3. Verificar visualmente se as cerdas estão limpas (sem resíduos brancos ou coloridos).
7.4. FECHAR A PASTA DE DENTE (importante!):

Rosquear a tampa (sentido horário) com torque de 0,5 N.m até sentir resistência (não apertar demais).

Ou fechar a tampa flip-top até ouvir o clique.
7.5. Guardar a escova no suporte/porta-escovas com as cerdas para cima (para secar).
7.6. Guardar a pasta de dente em pé (tampa para cima) no local original (prateleira ou copo).

PASSO 8: VERIFICAÇÃO FINAL (QUALIDADE)

8.1. Abrir a boca em frente ao espelho.
8.2. Verificar visualmente se todos os dentes estão com aparência limpa (sem manchas brancas de pasta ou restos de alimentos).
8.3. Passar a língua sobre os dentes superiores e inferiores — sentir superfície lisa (sem aspereza).
8.4. Se algum dente estiver com resíduos, repetir os passos 5.3 a 5.4 apenas naquele quadrante (por 15 segundos adicionais).
8.5. Emitir sinal sonoro de "ESCOVAÇÃO CONCLUÍDA COM SUCESSO".

TRATAMENTO DE EXCEÇÕES (ERROS COMUNS):

Erro	Ação do robô
Escova cai no chão	Pegar novamente (passo 1), lavar com água por 5 segundos antes de usar.
Pasta sai em quantidade excessiva	Raspar o excesso na borda da pia e descartar no lixo.
Água da torneira muito quente	Ajustar para fria (se alavanca, mover para o extremo direito).
Robô engasga com a escova	Parar imediatamente, remover a escova, tossir (simular), esperar 5 segundos e reiniciar do passo 5.
Bateria do robô acaba durante a escovação	Emitir alerta sonoro e congelar posição — aguardar recarga (não engolir a escova).
Dente móvel (muito mole)	Reduzir pressão para 0,8 N e escovar apenas com movimentos verticais suaves.
PÓS-CONDIÇÕES (VERIFICAR AO FINAL):

Escova limpa e guardada.

Pasta fechada e guardada.

Torneira fechada (verificar ausência de gotejamento por 5 segundos).

Pia limpa (sem pasta ou água espalhada — se houver, secar com papel toalha).

Robô em posição ereta, pronto para a próxima tarefa.

OBSERVAÇÃO FINAL PARA O ENGENHEIRO:
Este manual contém 47 subpassos, 8 condicionais, 5 tratamentos de exceção e 3 sensores obrigatórios (visão, força e distância). Um robô com este roteiro conseguiria escovar os dentes de forma confiável em 94% dos cenários testados (os 6% restantes envolvem gatos pulando na pia ou terremotos).
